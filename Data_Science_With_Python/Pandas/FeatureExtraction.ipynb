```python
def extract_episodes(txt):
    check=False;
    data=""
    for i in txt:
       if i==")":
           break;
       else:
        if i=="(":
            check=True;
        if check==True:
            data=data+i.replace("(","");
            
    return data;
                
df["Title"].apply(extract_episodes)
                
```
