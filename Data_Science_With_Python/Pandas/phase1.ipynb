<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Finding Missing Data
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3)# this mean that atleast we should have 3 non null values in a row if the doesnot have so it will be removed
```

```python
df
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Filling the Missing Data
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
df.fillna(0)
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Fill missing values with different values for each column
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
values={"A":0,"B":100,"C":300,"D":400}
df.fillna(value=values)
```
