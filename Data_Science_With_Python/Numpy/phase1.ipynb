<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
  Creating Array from scratch
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
zeros=np.zeros((3,4))
print(zeros)


ones=np.ones((3,4))
print(ones)

full=np.full((2,2),7)
print("\n full",full)

random=np.random.random((2,4))
print("\n random",random)


sequence=np.arange(0,10,2)
print("\n sequence array",sequence)
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
  Array Reshaping
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
arr=np.arange(12)
print("Original array ",arr)

reshaped=arr.reshape((3,4))
print("\n Reshaped array ",reshaped)

flattened=reshaped.flatten() # it gives copy instead of original array
print("\n Flattened array ",flattened)

raveled=reshaped.ravel()()# it gives original array instead of copy
print("\n Raveled array ",raveled)

Transpose=reshaped.T
print("\n Transposed array ",Transpose)
```
