```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
print("=== Zomato Sales Analysis ===")
print("\n Sales data shape",sales_data.shape)
print("\n Sample data for first three restaurants: ",sales_data[0:3])
print("\n Sample data for first three restaurants without id: ",sales_data[0:3,1:])
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Total Sales per year
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
yearly_total=np.sum(sales_data[:,1:],axis=0)
print("yearly_total",yearly_total)
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Minimum Sales per restaurant
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
minimum_sales_per_rest=np.min(sales_data[:,1:],axis=1)
print(minimum_sales_per_rest)
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Maximum Sales per year
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
max_sales_per_year=np.max(sales_data[:,1:],axis=0)
print(max_sales_per_year)
```

<h2 style="text-align: center; font-family: Georgia, serif; letter-spacing: 3px; color: #fff; font-size: 1.6em; font-weight: 700; margin-bottom: 0;">
Average Sales per restaurant
</h2>
<hr style="border: none; height: 1px; background-color: #fff; width: 35%; margin: 6px auto;">

```python
avg_sales_per_rest=np.average(sales_data[:,1:],axis=1)
print(avg_sales_per_rest)
```

```python
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])
print("vector addition ",vector1+vector2)
print("vector multiplication ",vector1*vector2)
print("Dot product ",np.dot(vector1,vector2))
```
