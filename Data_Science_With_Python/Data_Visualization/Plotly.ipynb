```python
import seaborn as sns
import plotly.express as px

df = sns.load_dataset('tips')

fig = px.line(df, y='total_bill', title='Total Bill')
fig.show()
```

```python
bar_chart=df.groupby("day")['tip'].mean().reset_index()
fig=px.bar(bar_chart,x="day",y="tip",color_discrete_sequence=["orange"])
fig.show()
```

```python
scatter_chart=df.groupby("total_bill")['tip'].mean().reset_index()
fig=px.scatter(scatter_chart,x="total_bill",y="tip",color_discrete_sequence=["orange"])
fig.show()
```
