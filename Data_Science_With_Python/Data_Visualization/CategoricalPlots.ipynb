```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
```

**Count Plot**

**It simply counts how many times each category appears and shows it as bars.**


```python
sns.countplot(x="sex",hue="smoker",data=df)
```

**Bar Plot**

```python
sns.barplot(x="sex",y="tip",data=df)
```

```python
import numpy as np
sns.barplot(x="sex",y="total_bill",data=df,estimator=np.sum)
```

**Box Plot**

```python
sns.boxplot(x="total_bill",y="day",data=df)
```

**ViolinPlot**

```python
sns.violinplot(x="day",y="total_bill",data=df)
```

**Strip Plot**

```python
sns.stripplot(x="day",y="total_bill",data=df)
```

**Swarmplot**

```python
sns.swarmplot(x="day",y="total_bill",data=df)
```

```python
sns.violinplot(x="tip",y="day",data=df)
sns.swarmplot(x="tip",y="day",data=df)
```
