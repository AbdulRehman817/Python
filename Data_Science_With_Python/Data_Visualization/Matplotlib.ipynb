```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x=np.linspace(0,5,11)
x
y=x**2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X")
plt.ylabel("Y")
```

**Sub in Matplotlib**

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(x**2,y**2)
plt.subplot(2,2,4)\
    
plt.plot(x,y**4)
```

**Object Oriented Way**

```python
fig=plt.figure(figsize=(12,8)) #COMMENT: 12 is width and 8 is height
axis=fig.add_axes([0.1,0.1,0.8,0.8])#COMMENT: here 0.1 means left , bottm, width, height
axis.plot(x,y)
axis1=fig.add_axes([0.4,0.1,0.5,0.5])
axis1.plot(x,y)
```

**Types of plots**

**Saving plots**

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("myplot.png")
```
